```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import mediapipe as mp

# Read CSV
data = pd.read_csv("./Profundidad/test_3D.csv")

POSE_CONNECTIONS = list(mp.solutions.pose.POSE_CONNECTIONS)

frame = 200

fig = plt.figure(figsize=(8,8))
ax = fig.add_subplot(111, projection='3d')

# ---------------------------------------
# Read landmarks
# ---------------------------------------

points = {}

for i in range(33):

    X = data.loc[frame, f"X_{i}"]
    Y = data.loc[frame, f"Y_{i}"]
    Z = data.loc[frame, f"Z_{i}"]

    # Skip missing landmarks
    if np.isnan(X) or np.isnan(Y) or np.isnan(Z):
        continue

    # Make person upright
    Y = -Y

    points[i] = (X, Z, Y)

# ---------------------------------------
# Draw joints
# ---------------------------------------

for i, (x, z, y) in points.items():

    ax.scatter(
        x,
        z,
        y,
        color='green',
        s=60
    )

    # Optional landmark labels
    ax.text(
        x,
        z,
        y,
        str(i),
        fontsize=8
    )

# ---------------------------------------
# Draw bones
# ---------------------------------------

for start, end in POSE_CONNECTIONS:

    # Draw only if BOTH landmarks exist
    if start not in points or end not in points:
        continue

    x1, z1, y1 = points[start]
    x2, z2, y2 = points[end]

    ax.plot(
        [x1, x2],
        [z1, z2],
        [y1, y2],
        color='blue',
        linewidth=2
    )

# ---------------------------------------
# Axes
# ---------------------------------------

len(points)

if len(points) > 0:

    coords = np.array(list(points.values()))

    mins = coords.min(axis=0)
    maxs = coords.max(axis=0)

    center = (mins + maxs) / 2

    radius = np.max(maxs - mins)*2

    print(radius)
    print(center)

    ax.set_xlim(center[0]-radius, center[0]+radius)
    ax.set_ylim(center[1]-radius, center[1]+radius)
    ax.set_zlim(center[2]-radius, center[2]+radius)

ax.set_xlabel("X (m)")
ax.set_ylabel("Depth (m)")
ax.set_zlabel("Height (m)")

ax.view_init(elev=20, azim=-30)

plt.show()
```
